```python
import pandas as pd
data = {
    "Toss" : ["W","W","L","L","W","W","L"],
    "Temp" : ["HOT","MILD","HOT","HOT","COOL","MILD","COOL"],
    "Venue" : ["MUMBAI","DELHI","CHENNAI","MUMBAI","JAYPUR","CHENNAI","MUMBAI"],
    "Match" : ["w","w","l","l","w","w","l"]
}

df = pd.DataFrame(data)
print(df)
```

```python
import math
x = df.drop("Match",axis=1)
print("Features")
print(x)
y = df["Match"]
print("Target")
print(y)
```

```python
classes = y.unique()
print("Classes ")
print(classes)
```

```python
priors = {}
for c in classes:
  priors[c] = sum(y==c)/len(y)

print("Probabilities")
for c in classes:
  print("P(" + c + ") = " ,priors[c])
```

```python
def conditional_probability(feature,value,target_class):
  class_count = sum(y==target_class)
  count = 0
  for i in range(len(df)):
    if df.iloc[i]["Match"] == target_class and df.iloc[i][feature] == value:
      count += 1


  possible_values = df[feature].nunique()
  probability = (count + 1)/(class_count + possible_values)
  return probability
```

```python
def predict(row):
  probabilities = {}
  for c in classes:
    probability = priors[c]
    for feature in x.columns:
      value = row[feature]
      probability *= conditional_probability(feature,value,c)
    probabilities[c] = probability
  return probabilities

  sample_row = x.iloc[0]
  predicted_probabilities = predict(sample_row)

  print("Probabilities for sample row:")
  for c in predicted_probabilities:
    print("P(" + c + "|x) =", predicted_probabilities[c])
    prediction = max(
        probabilities,
        key=probabilities.get
    )

  return prediction
```

```python
new_match = pd.Series(
    {"Toss":"W","Temp":"HOT","Venue":"JAYPUR"}
)

prediction = predict(new_match)
print("Prediction:", prediction)
```

```python
predictions = []
for i in range(len(x)):
  prediction = predict(x.iloc[i])
  predictions.append(prediction)

print("Predictions:")
print(predictions)
```

```python
correct = 0
for actual , predicted in zip(y,predictions):
  if actual == predicted:
    correct += 1

accuracy = correct/len(y)
print("Predictions : ")
print(predictions)
print("Actual Values : ")
print(list(y))
print("Accuracy : " , accuracy * 100 ,"%")
```
